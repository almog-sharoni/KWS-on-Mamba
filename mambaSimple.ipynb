```python
# regular_mamba_test.py

import torch
import numpy as np
from mamba_ssm import Mamba as MambaSimple
# Define constants matching C implementation
STATE_DIM = 3
BATCH_SIZE = 1
SEQ_LENGTH = 2
INPUT_DIM = 4
OUTPUT_DIM = 3
FIXED_POINT_SHIFT = 4

# Initialize Mamba
mamba = MambaSimple(STATE_DIM, INPUT_DIM, OUTPUT_DIM, FIXED_POINT_SHIFT)
# mamba.init_weights()

# Define test input
input_tensor = torch.tensor([[[1, 2, 3, 4],
                              [5, 6, 7, 8]]], dtype=torch.float32)

# Forward pass
output_tensor = mamba.forward(input_tensor)

# Convert to fixed-point scaling
scaled_output = (output_tensor.numpy() * (2**FIXED_POINT_SHIFT)).astype(np.int8)

print("Scaled Output from Regular Mamba:")
print(scaled_output)
```
